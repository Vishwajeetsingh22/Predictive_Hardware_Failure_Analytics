```python
# ============================================================
# MEMBER 3
# EDA & STATISTICAL ANALYSIS
# MEMORY-EFFICIENT VERSION
# STEPS 22–56
# ============================================================

import pandas as pd
import numpy as np
import os
import sys
import gc

from IPython.display import display

# ============================================================
# SETTINGS
# ============================================================

INPUT_FILE = r"C:\Users\admin\OneDrive\Desktop\DS- Activity 1\backblaze_analytical_dataset.csv"

OUTPUT_FOLDER = "output/eda_results"

CHUNK_SIZE = 100_000

# Sample size used for operations that require
# storing observations such as quantiles/correlation.
SAMPLE_SIZE = 500_000

os.makedirs(
    OUTPUT_FOLDER,
    exist_ok=True
)

pd.set_option(
    "display.max_columns",
    None
)

pd.set_option(
    "display.max_rows",
    100
)

pd.set_option(
    "display.width",
    200
)


# ============================================================
# HELPER — JUPYTER + TERMINAL + CSV
# ============================================================

def show_and_save(
    data,
    filename,
    title
):

    print("\n" + "=" * 70)
    print(title)
    print("=" * 70)

    display(data)

    try:

        sys.__stdout__.write(
            "\n" +
            "=" * 70 +
            "\n" +
            title +
            "\n" +
            "=" * 70 +
            "\n" +
            data.to_string(
                index=False
            ) +
            "\n"
        )

        sys.__stdout__.flush()

    except Exception:
        pass

    filepath = os.path.join(
        OUTPUT_FOLDER,
        filename
    )

    data.to_csv(
        filepath,
        index=False
    )

    print(
        "\nSaved CSV:"
    )

    print(
        filepath
    )

    try:

        sys.__stdout__.write(
            f"\nSaved CSV: {filepath}\n"
        )

        sys.__stdout__.flush()

    except Exception:
        pass

    return data
```

```python
# ============================================================
# 1. LOAD DATASET
# ============================================================

INPUT_FILE = r"C:\Users\admin\OneDrive\Desktop\DS- Activity 1\backblaze_analytical_dataset.csv"

OUTPUT_FOLDER = "output"

os.makedirs(
    OUTPUT_FOLDER,
    exist_ok=True
)


print("=" * 60)
print("DATASET LOADING")
print("=" * 60)


df = pd.read_csv(
    INPUT_FILE,
    low_memory=False
)


# Save original dataset information

original_rows = df.shape[0]

original_columns = df.shape[1]


print(
    "Dataset loaded successfully."
)

print(
    "Rows    :",
    f"{original_rows:,}"
)

print(
    "Columns :",
    original_columns
)


# ============================================================
# MEMORY CHECK FUNCTION
# ============================================================

def show_memory_usage(
    dataframe,
    message="Memory usage"
):

    memory_mb = (
        dataframe.memory_usage(
            deep=True
        ).sum()
        / (1024 ** 2)
    )

    print(
        f"{message}: "
        f"{memory_mb:.2f} MB"
    )


show_memory_usage(
    df,
    "Initial DataFrame memory"
)
```

```python
# ============================================================
# 2. BASIC DATASET INSPECTION
# ============================================================

print("\n" + "=" * 60)
print("FIRST 5 ROWS")
print("=" * 60)

print(
    df.head()
)


print("\n" + "=" * 60)
print("COLUMN NAMES")
print("=" * 60)

print(
    df.columns.tolist()
)


print("\n" + "=" * 60)
print("DATA TYPES AND INFORMATION")
print("=" * 60)

df.info()
```

```python
# ============================================================
# 3. CHECK MISSING VALUES
# ============================================================

print("\n" + "=" * 60)
print("MISSING VALUES BEFORE PREPROCESSING")
print("=" * 60)


missing_values = df.isnull().sum()


print(
    missing_values
)


print(
    "\nMissing Value Percentage:"
)


missing_percentage = (
    df.isnull().sum()
    / len(df)
) * 100


print(
    missing_percentage.sort_values(
        ascending=False
    )
)
```

```python
# ============================================================
# 4. IDENTIFY NUMERICAL AND CATEGORICAL COLUMNS
# ============================================================

numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()


categorical_columns = df.select_dtypes(
    exclude=np.number
).columns.tolist()


print("\n" + "=" * 60)
print("NUMERICAL COLUMNS")
print("=" * 60)

print(
    numeric_columns
)


print("\n" + "=" * 60)
print("CATEGORICAL COLUMNS")
print("=" * 60)

print(
    categorical_columns
)
```

```python
# ============================================================
# 5. HANDLE MISSING VALUES
# ============================================================

print("\n" + "=" * 60)
print("HANDLING MISSING VALUES")
print("=" * 60)


# ------------------------------------------------------------
# Numerical columns
# Fill missing values using median
# ------------------------------------------------------------

for column in numeric_columns:

    if df[column].isnull().any():

        median_value = df[
            column
        ].median()

        df[column] = df[
            column
        ].fillna(
            median_value
        )


# ------------------------------------------------------------
# Categorical columns
# Fill missing values using mode
# ------------------------------------------------------------

for column in categorical_columns:

    if df[column].isnull().any():

        mode_values = df[
            column
        ].mode(
            dropna=True
        )

        if len(mode_values) > 0:

            df[column] = df[
                column
            ].fillna(
                mode_values.iloc[0]
            )

        else:

            df[column] = df[
                column
            ].fillna(
                "Unknown"
            )


print(
    "Total missing values after handling:",
    int(
        df.isnull().sum().sum()
    )
)
```

```python
# ============================================================
# 6. CHECK DUPLICATES
# ============================================================

print("\n" + "=" * 60)
print("DUPLICATE RECORDS")
print("=" * 60)


duplicates_before = (
    df.duplicated().sum()
)


print(
    "Duplicates found:",
    duplicates_before
)


if duplicates_before > 0:

    df = df.drop_duplicates(
        ignore_index=True
    )


duplicates_after = (
    df.duplicated().sum()
)


print(
    "Duplicates after removal:",
    duplicates_after
)


gc.collect()
```

```python
# ============================================================
# 7. IDENTIFY POSSIBLE TARGET COLUMN
# ============================================================

print("\n" + "=" * 60)
print("TARGET COLUMN DETECTION")
print("=" * 60)


possible_targets = [
    "failure",
    "failed",
    "disk_failure",
    "disk_failed",
    "failure_status",
    "target",
    "label"
]


target_column = None


# ------------------------------------------------------------
# Exact match
# ------------------------------------------------------------

for column in possible_targets:

    if column in df.columns:

        target_column = column

        break


# ------------------------------------------------------------
# Case-insensitive match
# ------------------------------------------------------------

if target_column is None:

    lower_column_mapping = {
        column.lower(): column
        for column in df.columns
    }


    for target_name in possible_targets:

        if (
            target_name.lower()
            in lower_column_mapping
        ):

            target_column = (
                lower_column_mapping[
                    target_name.lower()
                ]
            )

            break


if target_column is not None:

    print(
        "Target column detected:",
        target_column
    )

else:

    print(
        "No automatic target column detected."
    )
```

```python
# ============================================================
# 8. OUTLIER DETECTION USING IQR
# ============================================================

print("\n" + "=" * 60)
print("OUTLIER DETECTION")
print("=" * 60)


numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()


outlier_columns = (
    numeric_columns.copy()
)


if (
    target_column is not None
    and target_column in outlier_columns
):

    outlier_columns.remove(
        target_column
    )


outlier_summary = {}


for column in outlier_columns:

    Q1 = df[
        column
    ].quantile(
        0.25
    )


    Q3 = df[
        column
    ].quantile(
        0.75
    )


    IQR = Q3 - Q1


    lower_bound = (
        Q1 - 1.5 * IQR
    )


    upper_bound = (
        Q3 + 1.5 * IQR
    )


    outlier_count = (
        (
            df[column] < lower_bound
        )
        |
        (
            df[column] > upper_bound
        )
    ).sum()


    outlier_summary[
        column
    ] = int(
        outlier_count
    )


    print(
        f"{column}: "
        f"{outlier_count:,} outliers"
    )
```

```python
# ============================================================
# 9. OUTLIER HANDLING USING IQR CAPPING
# ============================================================

print("\n" + "=" * 60)
print("HANDLING OUTLIERS")
print("=" * 60)


for column in outlier_columns:

    Q1 = df[
        column
    ].quantile(
        0.25
    )


    Q3 = df[
        column
    ].quantile(
        0.75
    )


    IQR = Q3 - Q1


    lower_bound = (
        Q1 - 1.5 * IQR
    )


    upper_bound = (
        Q3 + 1.5 * IQR
    )


    df[column] = df[
        column
    ].clip(
        lower=lower_bound,
        upper=upper_bound
    )


print(
    "Outlier capping completed."
)
```

```python
# ============================================================
# 10. CATEGORICAL DATA ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("CATEGORICAL DATA ANALYSIS")
print("=" * 60)


categorical_columns = df.select_dtypes(
    exclude=np.number
).columns.tolist()


if len(categorical_columns) == 0:

    print(
        "No categorical columns found."
    )

else:

    for column in categorical_columns:

        print(
            "\nColumn:",
            column
        )


        print(
            "Unique values:",
            df[column].nunique(
                dropna=False
            )
        )


        print(
            df[column]
            .value_counts()
            .head(10)
        )
```

```python
# ============================================================
# 11. REMOVE HIGH-CARDINALITY ID COLUMNS
# ============================================================

print("\n" + "=" * 60)
print("HIGH-CARDINALITY ID CHECK")
print("=" * 60)


categorical_columns = df.select_dtypes(
    exclude=np.number
).columns.tolist()


id_columns = []


for column in categorical_columns:

    column_lower = column.lower()


    is_id_name = (

        column_lower in [
            "id",
            "device_id",
            "disk_id",
            "serial",
            "serial_number",
            "serialnumber"
        ]

        or column_lower.endswith(
            "_id"
        )

        or "serial" in column_lower
    )


    if is_id_name:

        unique_count = df[
            column
        ].nunique(
            dropna=False
        )


        unique_ratio = (
            unique_count
            / len(df)
        )


        print(
            f"{column}: "
            f"{unique_count:,} unique values "
            f"({unique_ratio:.2%} unique)"
        )


        if unique_ratio > 0.50:

            id_columns.append(
                column
            )


if len(id_columns) > 0:

    print(
        "\nIdentifier columns removed:"
    )

    print(
        id_columns
    )


    df = df.drop(
        columns=id_columns
    )


else:

    print(
        "\nNo high-cardinality "
        "identifier columns detected."
    )


# Refresh column lists

numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()


categorical_columns = df.select_dtypes(
    exclude=np.number
).columns.tolist()


print(
    "\nRemaining categorical columns:"
)

print(
    categorical_columns
)


gc.collect()
```

```python
# ============================================================
# 12. MEMORY-SAFE CATEGORICAL ENCODING
# ============================================================

print("\n" + "=" * 60)
print("MEMORY-SAFE CATEGORICAL ENCODING")
print("=" * 60)


# Maximum categories allowed
# for one-hot encoding

MAX_CATEGORIES = 20


columns_to_encode = []

columns_not_encoded = []


for column in categorical_columns:

    unique_count = df[
        column
    ].nunique(
        dropna=False
    )


    if unique_count <= MAX_CATEGORIES:

        columns_to_encode.append(
            column
        )

    else:

        columns_not_encoded.append(
            column
        )


print(
    "\nColumns selected for encoding:"
)

print(
    columns_to_encode
)


print(
    "\nHigh-cardinality columns NOT encoded:"
)

print(
    columns_not_encoded
)


# ------------------------------------------------------------
# Convert selected columns to category
# ------------------------------------------------------------

for column in columns_to_encode:

    df[column] = df[
        column
    ].astype(
        "category"
    )


# ------------------------------------------------------------
# Sparse one-hot encoding
# ------------------------------------------------------------

if len(columns_to_encode) > 0:

    df = pd.get_dummies(
        df,
        columns=columns_to_encode,
        drop_first=True,
        dtype="int8",
        sparse=True
    )


    print(
        "\nCategorical encoding completed."
    )

else:

    print(
        "\nNo categorical columns "
        "required encoding."
    )


print(
    "\nDataset shape after encoding:"
)

print(
    df.shape
)


show_memory_usage(
    df,
    "Memory after encoding"
)


gc.collect()
```

```python
# ============================================================
# 13. CONVERT BOOLEAN COLUMNS
# ============================================================

print("\n" + "=" * 60)
print("CONVERT BOOLEAN COLUMNS")
print("=" * 60)


boolean_columns = df.select_dtypes(
    include="bool"
).columns.tolist()


if len(boolean_columns) > 0:

    for column in boolean_columns:

        df[column] = df[
            column
        ].astype(
            "int8"
        )


    print(
        f"{len(boolean_columns)} "
        "boolean columns converted."
    )

else:

    print(
        "No boolean columns found."
    )
```

```python
# ============================================================
# 14. SKEWNESS CHECK
# ============================================================

print("\n" + "=" * 60)
print("SKEWNESS OF NUMERICAL FEATURES")
print("=" * 60)


numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()


# Do not calculate skewness on target

skewness_columns = (
    numeric_columns.copy()
)


if (
    target_column is not None
    and target_column in skewness_columns
):

    skewness_columns.remove(
        target_column
    )


if len(skewness_columns) > 0:

    skewness = df[
        skewness_columns
    ].skew()


    print(
        skewness.sort_values(
            ascending=False
        )
    )

else:

    print(
        "No numerical features "
        "available."
    )
```

```python
# ============================================================
# 15. LOG TRANSFORMATION
# ============================================================

print("\n" + "=" * 60)
print("LOG TRANSFORMATION")
print("=" * 60)


transformation_columns = (
    skewness_columns.copy()
)


for column in transformation_columns:

    try:

        skew_value = df[
            column
        ].skew()


        minimum_value = df[
            column
        ].min()


        if (
            skew_value > 1
            and minimum_value >= 0
        ):

            df[column] = np.log1p(
                df[column]
            )


            print(
                f"Log transformation applied: "
                f"{column}"
            )


    except Exception as error:

        print(
            f"Skipped {column}: "
            f"{error}"
        )


print(
    "Transformation completed."
)
```

```python
# ============================================================
# 16. STANDARDIZATION
# ============================================================

print("\n" + "=" * 60)
print("STANDARDIZATION")
print("=" * 60)


# ------------------------------------------------------------
# IMPORTANT:
# Only standardize ORIGINAL CONTINUOUS NUMERICAL FEATURES.
#
# Do NOT standardize one-hot encoded columns.
# Do NOT standardize the target column.
# ------------------------------------------------------------


numeric_columns = df.select_dtypes(
    include=np.number
).columns.tolist()


# Columns created by one-hot encoding
# usually contain "__" in pandas naming,
# but we use the actual list created above
# to be completely safe.


encoded_columns = []


for column in df.columns:

    # Identify sparse dummy columns

    if hasattr(
        df[column].dtype,
        "subtype"
    ):

        if (
            df[column].dtype.subtype
            == np.dtype("int8")
        ):

            encoded_columns.append(
                column
            )


# ------------------------------------------------------------
# Better identification:
# exclude all columns that were created
# during categorical encoding.
# ------------------------------------------------------------

encoded_columns = []


if len(columns_to_encode) > 0:

    for original_column in columns_to_encode:

        prefix = (
            str(original_column)
            + "_"
        )


        for column in df.columns:

            if str(column).startswith(
                prefix
            ):

                encoded_columns.append(
                    column
                )


# Remove duplicates

encoded_columns = list(
    dict.fromkeys(
        encoded_columns
    )
)


# ------------------------------------------------------------
# Select scaling columns
# ------------------------------------------------------------

scaling_columns = []


for column in numeric_columns:

    # Skip target

    if (
        target_column is not None
        and column == target_column
    ):

        continue


    # Skip encoded columns

    if column in encoded_columns:

        continue


    # Only continuous numerical
    # columns are standardized

    scaling_columns.append(
        column
    )


print(
    "\nNumerical columns:",
    len(numeric_columns)
)


print(
    "Encoded columns excluded:",
    len(encoded_columns)
)


print(
    "Columns selected for scaling:",
    len(scaling_columns)
)


print(
    "\nScaling columns:"
)

print(
    scaling_columns
)


# ------------------------------------------------------------
# Standardize safely
# ------------------------------------------------------------

if len(scaling_columns) > 0:

    print(
        "\nConverting scaling columns "
        "to float32..."
    )


    # Explicitly convert these columns
    # to floating-point BEFORE assigning
    # standardized values.

    for column in scaling_columns:

        df[column] = pd.to_numeric(
            df[column],
            errors="coerce"
        ).astype(
            np.float32
        )


    # --------------------------------------------------------
    # StandardScaler
    # --------------------------------------------------------

    scaler = StandardScaler()


    print(
        "Calculating standardized values..."
    )


    scaled_array = scaler.fit_transform(
        df[
            scaling_columns
        ]
    )


    # Convert output to float32

    scaled_array = scaled_array.astype(
        np.float32
    )


    # --------------------------------------------------------
    # SAFE COLUMN-BY-COLUMN ASSIGNMENT
    # --------------------------------------------------------

    for index, column in enumerate(
        scaling_columns
    ):

        df[column] = scaled_array[
            :, index
        ]


    # Delete temporary objects

    del scaled_array

    del scaler

    gc.collect()


    print(
        "\nStandardization completed successfully."
    )


else:

    print(
        "\nNo continuous numerical "
        "features available for scaling."
    )
```
